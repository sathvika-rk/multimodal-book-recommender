```python
from google import genai
from PIL import Image
from google.colab import userdata
```

```python
api_key=userdata.get('GEMINI_API_KEY')
client=genai.Client(api_key=api_key)
image_path='/content/WhatsApp Image 2026-06-23 at 02.39.55.jpeg'
book_image=Image.open(image_path)
book_image.thumbnail((300,300))
display(book_image)
```

```python
import time
time.sleep(2)
prompt = """
You are an expert librarian and data extraction assistant.
Analyze the provided image of books.
Extract the title and author for every book clearly visible in the image.
Output your response ONLY as a valid JSON array of objects.
The format must be exactly: [{"title": "Book Title", "author": "Author Name"}, ...]
Do not include any conversational text or formatting outside the JSON.
"""
response=client.models.generate_content(model='gemini-2.5-flash',contents=[prompt,book_image])
print(response.text)
```
